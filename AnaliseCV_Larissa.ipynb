```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python

im_bgr = cv2.imread('ruareal.jpeg') #em bgr
im = imagem_original = cv2.cvtColor(im_bgr, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(8, 6))
plt.imshow(im)
plt.axis('off')
plt.show()
```

```python
print(im[150, 200])
```

```python
im_brilho = cv2.convertScaleAbs(im, alpha=1, beta=50)
plt.figure(figsize=(8, 6))
plt.imshow(im_brilho)
plt.axis('off')
plt.show()
```

```python
im_contr = cv2.convertScaleAbs(im, alpha=1.5, beta=0)
plt.imshow(im_contr)
plt.axis('off')
plt.show()
```

```python
# 1. Aplicando a matemática (ajuste os valores de alpha e beta se quiser testar!)


# 2. Criando o painel com 3 espaços lado a lado
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# Espaço 0: Original
ax[0].imshow(im)
ax[0].set_title('Original')
ax[0].axis('off')

# Espaço 1: Brilho
ax[1].imshow(im_brilho)
ax[1].set_title('Brilho Ajustado (beta=50)')
ax[1].axis('off')

# Espaço 2: Contraste
ax[2].imshow(im_contr)
ax[2].set_title('Alto Contraste (alpha=1.5)')
ax[2].axis('off')

# Mostra o resultado final
plt.show()
```

```python
spl_im = cv2.split(im)

fig, ax = plt.subplots(1, 4, figsize=(20,10))


ax[0].imshow(spl_im[0], cmap='Reds')
ax[0].axis('off')
ax[0].set_title('Vermelho')

ax[1].imshow(spl_im[1], cmap='Greens')
ax[1].axis('off')
ax[1].set_title('Verde')

ax[2].imshow(spl_im[2], cmap='Blues')
ax[2].axis('off')
ax[2].set_title('Azul')

ax[3].imshow(im)
ax[3].axis('off')
ax[3].set_title('Original')



```

```python
matriz20= np.ones((20,20), dtype=np.float32)/400
matriz8= np.ones((8,8), dtype=np.float32)/64
blurred_im8 = cv2.filter2D(im, -1, matriz8)
blurred_im20 = cv2.filter2D(im, -1, matriz20)
fig, ax = plt.subplots(1, 2, figsize=(20,10))
ax[1].imshow(blurred_im8)
ax[1].axis('off')
ax[1].set_title('Blur com matriz 8x8')
ax[0].set_title('Original')
ax[0].imshow(im)
ax[0].axis('off')


```

```python
cal_bgr = cv2.imread('ruareal.jpeg') #em bgr
cal = cv2.cvtColor(cal_bgr, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 6))
plt.imshow(cal)
plt.axis('off')
plt.show()
```

```python


Mx = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
My = np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]])

im_Mx = cv2.filter2D(cal, -1, Mx)
im_My = cv2.filter2D(cal, -1, My)
fig, ax = plt.subplots(1, 3, figsize=(20,10))
ax[1].imshow(im_Mx)
ax[1].axis('off')
ax[1].set_title('Convolução eixo x')
ax[0].set_title('Original')
ax[0].imshow(cal)
ax[0].axis('off')
ax[2].imshow(im_My)
ax[2].axis('off')
ax[2].set_title('Convolução eixo y')
```

```python

matriz8= np.ones((8,8), dtype=np.float32)/64
blurred_cal8 = cv2.filter2D(cal, -1, matriz8)

cal_Mx = cv2.filter2D(blurred_cal8, -1, Mx)
cal_My = cv2.filter2D(blurred_cal8, -1, My)
fig, ax = plt.subplots(1, 3, figsize=(20,10))
ax[0].set_title('Imagem borrada 8x8')
ax[0].imshow(blurred_cal8)
ax[0].axis('off')
ax[1].imshow(cal_Mx)
ax[1].axis('off')
ax[1].set_title('Convolução eixo x')
ax[2].imshow(cal_My)
ax[2].axis('off')
ax[2].set_title('Convolução eixo y')
```
